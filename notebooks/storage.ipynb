# Checkpoints Storage
The purpose of this notebook is to present how Torch Tweak stores and loads checkpoints.

Basically, storing and loading tuned models or pipelines consists of two parts:
- checkpointing mechanism - which uses torch `state_dict` to dump and load weights
- storage mechanism - which is used to write the resulting artifacts to disk.

This notebook presents the storage mechanism. The checkpointing is presented in `saving_loading_checkpoints.ipynb` notebook.

```python
%%capture
%cd ..
```

```python
import os
from logging import basicConfig

import torch
import torch.nn as nn
from torch_tweak.torch import LocalTorchStorage
from torch_tweak.torch.backend import TorchInductorBackend, OpenVINOBackend
from torch_tweak.torch import save, tune, LocalTorchStorage, OneBackendStrategy, TunableModule
```

```python
log_level = os.environ.get("TORCH_TWEAK_LOG_LEVEL", "INFO")
basicConfig(level=log_level, format="%(asctime)s - %(levelname)s - %(message)s", force=True)
device = torch.device("xpu" if torch.xpu.is_available() else "cpu")
```

# Introduction

First, let's create a fake `state_dict` which will imitate real torch module state dict.

```python
state_dict = {
    "layer": torch.randn(2, 2),
}

state_dict
```

Let's save a fake `state_dict` to a checkpoint.

```python
storage = LocalTorchStorage(base_folder="notebooks/checkpoints", remove_checkpoint_after_tune=True)
storage.save("demo-storage.tt", state_dict)
```

Let's see stored artifacts.

```python
!find notebooks/checkpoints/demo-storage[._]* -type f
```

There are two files:
- the checkpoints .tt file
- the sha sums file

```python
!cat notebooks/checkpoints/demo-storage_sha256_sums.txt
```

The `demo-storage.tt` checkpoint is actually a zip archive with all artifacts. When we do `load` for the first time the checkpoint is 
decompressed.


Let's try to `load` the checkpoint.

```python
loaded_state_dict = storage.load("demo-storage.tt")
loaded_state_dict
```

Let's see resulting folder structure.

```python
!tree -L 1 notebooks/checkpoints/demo-storage/
```

On the first load the sha sums are verified. On subsequent loads, to speed up the process, 
the original sha sums file is compared against the one in the decompressed folder. If they match checkpoint is loaded from decompressed folder and sums are not verified. However if they differ, the checkpoint is re-decompressed and again hash sums are verified.

# Storage with tuning

Let's create simple model and tune it.

```python
class SimpleNet(nn.Module):
    def __init__(self, input_size, hidden_size, num_classes):
        super(SimpleNet, self).__init__()
        self.fc1 = nn.Linear(input_size, hidden_size)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_size, hidden_size)
        self.relu = nn.ReLU()
        self.fc3 = nn.Linear(hidden_size, num_classes)

    def forward(self, x):
        out = self.fc1(x)
        out = self.relu(out)
        out = self.fc2(out)
        out = self.relu(out)
        out = self.fc3(out)
        return out
```

```python
net = SimpleNet(5, 10, 2).to(device)
tune_data = torch.randn(5, device=device)
```

Let's wrap modules.

```python
net.fc1 = TunableModule(net.fc1, "fc1", strategy=OneBackendStrategy(backend=TorchInductorBackend()))
net.fc2 = TunableModule(net.fc2, "fc2", strategy=OneBackendStrategy(backend=OpenVINOBackend()))
net.fc3 = TunableModule(net.fc3, "fc3", strategy=OneBackendStrategy(backend=OpenVINOBackend()))
```

Let's tune the model.

```python
tune(net, tune_data, batch_sizes=[1, 2], dry_run=False)
```

Let's save it.

```python
storage = LocalTorchStorage(
    base_folder="notebooks/checkpoints",
    remove_checkpoint_after_tune=False,  # this flag will leave uncompressed checkpoint after tuning
)
save(net, "demo-tuned.tt", storage=storage)
```

Let's inspect the checkpoint folder.

```python
!tree -L 2 notebooks/checkpoints/demo-tuned/
```

It contains:
- sha sums file
- `state_dict.pt` which will have torch module weights
- two `openvino` folders - since we have two modules tuned with `OpenVINOBackend`

If you would like to store checkpoints in a cloud, you can override LocalTorchStorage with a custom storage.
